```python
cat_cols = df.select_dtypes(exclude='number').columns

for col in cat_cols:
    df[col].fillna(df[col].mode()[0], inplace=True)
```

```python
df['Popularity'].fillna(df['Popularity'].median(), inplace=True)
```

```python
#histogram
plt.figure(figsize=(8,5))
sns.histplot(df["Vote_Average"], bins=20)
plt.title("Distribution of Movie Ratings")
plt.show()
```

```python
sns.histplot(df["Popularity"], bins=30)
```

```python
top = df.sort_values("Popularity",ascending=False).head(10)

sns.barplot(x="Popularity",y="Title",data=top)
```

```python
# Convert text columns to actual numbers
df['Vote_Count'] = pd.to_numeric(df['Vote_Count'], errors='coerce')
df['Vote_Average'] = pd.to_numeric(df['Vote_Average'], errors='coerce')
sns.heatmap(
    df[["Popularity","Vote_Count","Vote_Average"]].corr(),
    annot=True,
    cmap="Blues"
)
```

```python
sns.scatterplot(
    x="Vote_Count",
    y="Popularity",
    data=df
)
```
